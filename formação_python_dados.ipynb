````python
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt

def criar_banner(titulo, subtitulo, arquivo_saida, cor_fundo='#6a1b9a'):
    # Criar uma imagem com fundo roxo
    img = Image.new('RGB', (1200, 600), color=cor_fundo)
    draw = ImageDraw.Draw(img)

    # Tentar carregar uma fonte, ou usar a padrão
    try:
        font_title = ImageFont.truetype('/usr/share/fonts/truetype/liberation/LiberationSans-Bold.ttf', 80)
        font_sub = ImageFont.truetype('/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf', 40)
    except:
        font_title = ImageFont.load_default()
        font_sub = ImageFont.load_default()

    # Desenhar textos
    draw.text((100, 200), titulo, fill='white', font=font_title)
    draw.text((100, 320), subtitulo, fill='#e1bee7', font=font_sub)
    draw.rectangle([100, 400, 300, 410], fill='#9c27b0') # Linha decorativa

    img.save(arquivo_saida)
    return img

# Gerando a Capa do Curso Atualizada
capa = criar_banner('PYTHON PARA', 'ANÁLISE DE DADOS JÚNIOR', 'capa_curso.png')
plt.figure(figsize=(10,5))
plt.imshow(capa)
plt.axis('off')
plt.title('Prévia: Capa do Notebook')
plt.show()
````

***tratamento de erro***

````python
# Criamos a lista de fábricas
fabricas = ["local", "fora", "interno"]

# Pedimos para o usuário digitar a fábrica
digitado = input("digite a fabrica: ")

# IMPORTANTE: Verificamos se o item existe na lista ANTES de usar o .index()
# Isso evita o erro 'ValueError' caso o usuário digite algo que não existe
if digitado in fabricas:
    # O .index() retorna a posição numérica do item na lista
    index_nome = fabricas.index(digitado)
    print(f"A fábrica {digitado} está na posição {index_nome}")
else:
    print("Essa fábrica não existe na lista, por isso não tem uma posição (index).")
````

<h2 style="color:purple;">Tratamento de Erros: Try e Except</h2>

````python
# O bloco 'try' (tentar) envolve o código que pode gerar algum erro
try:
    # Solicitamos dois números para realizar uma divisão
    numero_um = int(input("Digite o numerador: "))
    numero_dois = int(input("Digite o denominador: "))

    resultado = numero_um / numero_dois
    print(f"O resultado da divisão é: {resultado}")

# O 'except' captura o erro específico e executa um plano B em vez de fechar o programa
except ValueError:
    # Este bloco executa se o usuário digitar um texto onde deveria ser número
    print("Erro: Você precisa digitar um número inteiro válido!")

except ZeroDivisionError:
    # Este bloco executa se o usuário tentar dividir por zero (o que é impossível na matemática)
    print("Erro: Não é possível dividir um número por zero!")

except Exception as erro:
    # Este é um 'coringa' que captura qualquer outro erro desconhecido
    print(f"Ocorreu um erro inesperado: {erro}")

finally:
    # O bloco 'finally' sempre será executado, independentemente de ter havido erro ou não
    print("Fim da operação de tratamento de erro.")
````

<h2 style="color:purple;">Diferença entre For e While</h2>

No Python, temos duas formas principais de repetir tarefas:

1. **Loop `for` (Para cada...):**
   - **Quando usar:** Quando você sabe exatamente o fim da repetição ou quer percorrer uma lista do início ao fim.
   - **Exemplo:** Percorrer todos os nomes em uma lista de convidados.
   - **Vantagem:** É mais limpo e difícil de gerar um "loop infinito".

2. **Loop `while` (Enquanto...):**
   - **Quando usar:** Quando você **não sabe** quantas vezes precisará repetir, mas sabe que deve continuar enquanto uma condição for verdadeira.
   - **Exemplo:** Pedir uma senha até que o usuário digite a correta.
   - **Cuidado:** Se você esquecer de atualizar a variável (como o `qtd_vezes = qtd_vezes + 1`), o programa ficará rodando para sempre!

````python
# 1. Primeiro, criamos uma lista vazia para guardar os itens
lista_gigante = []

# 2. Usamos o 'for' com 'range' para repetir exatamente 200.000 vezes
# O 'i' será o número de cada volta (de 0 até 199.999)
for i in range(200000):
    # 3. Adicionamos o número atual à nossa lista
    lista_gigante.append(i)

# 4. Verificamos o tamanho final da lista para confirmar
print("Quantidade de itens na lista:", len(lista_gigante))

# 5. Mostramos apenas os 5 primeiros para não travar a tela
print("Primeiros itens:", lista_gigante[:5])
````

````python
# 1. Criamos uma lista vazia para o exemplo com while
lista_while = []

# 2. No while, precisamos criar a variável de contagem manualmente
i = 0

# 3. O loop rodará enquanto 'i' for menor que 200.000
while i < 200000:
    # 4. Adicionamos o valor atual de 'i' na lista
    lista_while.append(i)

    # 5. IMPORTANTE: Somamos 1 ao 'i' para a próxima volta
    # Se esquecer esta linha, o loop fica infinito!
    i = i + 1

# 6. Verificamos o resultado
print("Quantidade final usando while:", len(lista_while))
print("Último item adicionado:", lista_while[-1])
````

<h2 style="color:purple;">Resumo Visual: For vs While</h2>

| Característica | Loop **For** | Loop **While** |
| :--- | :--- | :--- |
| **Quando usar** | Quando sabemos o limite (ex: 200 mil vezes) | Enquanto uma condição for real (ex: até acertar a senha) |
| **Controle** | Automático (o `range` cuida do contador) | Manual (você precisa somar `+1` na variável) |
| **Segurança** | Mais seguro contra travamentos | Risco de loop infinito se esquecer o incremento |

<h2 style="color:purple;">Parâmetros de Função: As Entradas de Dados</h2>

Parâmetros são variáveis que definimos dentro dos parênteses da função para que ela possa receber valores externos.

### Tipos Comuns de Parâmetros:
1. **Parâmetros Obrigatórios:** Você deve passar o valor, senão o código dá erro.
2. **Parâmetros com Valor Padrão (Default):** Se você não passar nada, a função usa um valor pré-definido.

````python
# 1. Função com parâmetros OBRIGATÓRIOS
def calcular_imc(peso, altura):
    # Esta função precisa de 2 dados para funcionar
    imc = peso / (altura ** 2)
    return f"Seu IMC é {imc:.2f}"

# 2. Função com parâmetro PADRÃO (default)
def configurar_usuario(nome, status="Ativo"):
    # Se não informarmos o status, ele será "Ativo" por padrão
    print(f"Usuário: {nome} | Status: {status}")

# --- Testando os códigos ---

# Chamando IMC (precisa dos dois valores)
print(calcular_imc(80, 1.80))

# Chamando Usuário (apenas um valor, o outro é padrão)
configurar_usuario("Fabiano")

# Chamando Usuário (mudando o padrão)
configurar_usuario("Ana", "Inativo")
````

<h2 style="color:purple;">Tipos de Parâmetros e Tipagem de Código</h2>

No Python, podemos ser bem específicos sobre como nossa função recebe dados:

1. **Parâmetros Posicionais (Obrigatórios):** Devem ser passados na ordem correta.
2. **Parâmetros Nomeados:** Quando você diz qual variável recebe qual valor ao chamar a função.
3. **Type Hinting (Dica de Tipo):** Uma boa prática para indicar se a função espera um texto (`str`), número (`int`), etc.

````python
# Usando Dicas de Tipo (Type Hints) para deixar o código profissional
def registrar_venda(produto: str, quantidade: int, valor_unitario: float):
    total = quantidade * valor_unitario
    return f"Venda de {produto}: Total R$ {total:.2f}"

# Chamada Posicional (segue a ordem)
print(registrar_venda("Teclado", 2, 150.0))

# Chamada Nomeada (ordem não importa se você der o nome)
print(registrar_venda(valor_unitario=50.0, produto="Mouse", quantidade=1))
````

<h2 style="color:purple;">Por que usar o Return?</h2>

O `return` serve para que a função **devolva** um dado para quem a chamou. Sem ele, o resultado do cálculo fica 'preso' dentro da função e se perde.

````python
# Exemplo 1: Função SEM return (O valor se perde)
def somar_sem_retorno(n1, n2):
    resultado = n1 + n2
    # Aqui o resultado só existe dentro da função.
    # Se tentarmos usar fora, não conseguiremos.

teste1 = somar_sem_retorno(10, 10)
print(f"Tentando usar o resultado sem return: {teste1}") # Vai aparecer 'None'

# Exemplo 2: Função COM return (O valor é entregue)
def somar_com_retorno(n1, n2):
    resultado = n1 + n2
    return resultado # Aqui enviamos o valor para fora da função

teste2 = somar_com_retorno(10, 10)
print(f"Resultado guardado com sucesso: {teste2}") # Agora o 20 está salvo na variável!

# Exemplo 3: Usando o retorno em outro cálculo
# Como o teste2 vale 20, posso continuar a matemática
print(f"O triplo do resultado é: {teste2 * 3}")
````

<h2 style="color:purple;">Exemplo Prático: Cálculo de Desconto</h2>

````python
# Esta função processa o valor e RETORNA apenas o desconto calculado
def calcular_desconto(valor_total: float) -> float:
    # Se o valor for maior que 100, damos 10% de desconto
    if valor_total > 100:
        return valor_total * 0.10
    else:
        return valor_total * 0.05

# Chamamos a função e guardamos o retorno
meu_valor = 150.0
desconto_recebido = calcular_desconto(meu_valor)

# Agora usamos o retorno para calcular o preço final
preco_final = meu_valor - desconto_recebido

print(f"Valor Original: R${meu_valor}")
print(f"Desconto ganho: R${desconto_recebido}")
print(f"Preço a pagar: R${preco_final}")
````

<h2 style="color:purple;">Exercício: Removendo Excessos com Funções</h2>

Nesta etapa, criaremos uma função que automatiza o corte de convidados excedentes utilizando o conceito de **fatiamento de listas** e **retorno de valores**.

````python
# Definimos a função que remove o excesso
def remover_excesso(lista_completa, limite):
    # Verificamos se o tamanho da lista é maior que o limite permitido
    if len(lista_completa) > limite:
        print(f"⚠️ Atenção: Removendo {len(lista_completa) - limite} convidados excedentes...")
        # Retornamos a lista fatiada do início (0) até o limite
        return lista_completa[:limite]
    else:
        # Se não houver excesso, retornamos a lista original intacta
        return lista_completa

# --- Testando a função na prática ---

lista_atual = ["Joana", "Paula", "Jessica", "Ronei", "Felix", "Pedro"]
limite_salao = 4

# Chamamos a função e GUARDAMOS a lista corrigida em uma nova variável
lista_final = remover_excesso(lista_atual, limite_salao)

print(f"Lista original: {lista_atual}")
print(f"Lista final (dentro da capacidade): {lista_final}")
````

<h2 style="color:purple;">Exercício: Remoção Dinâmica com While em Funções</h2>

Nesta abordagem, usamos o loop `while` para remover o último item da lista repetidamente até que o tamanho da lista seja igual ao limite permitido.

````python
# Criamos a função que utiliza o loop while para controlar o tamanho
def reduzir_lista_com_while(lista_convidados, limite):
    # Criamos uma cópia da lista para não alterar a original permanentemente por acidente
    lista_trabalho = lista_convidados.copy()

    # ENQUANTO o tamanho da lista for maior que o limite...
    while len(lista_trabalho) > limite:
        # O comando .pop() remove o último item da lista e o 'guarda' em uma variável
        removido = lista_trabalho.pop()
        print(f"🚫 Removendo: {removido} (Acima da capacidade)")

    # Quando o loop termina, significa que a lista está no tamanho certo
    return lista_trabalho

# --- Testando a execução ---

meus_convidados = ["Joana", "Paula", "Jessica", "Ronei", "Felix", "Pedro"]
limite_maximo = 3

# Chamamos a função e recebemos a lista filtrada
lista_atualizada = reduzir_lista_com_while(meus_convidados, limite_maximo)

print(f"\nLista Final Aprovada: {lista_atualizada}")
````
